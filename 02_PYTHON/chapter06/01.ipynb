- \_\_name\_\_: 모듈 이름

### 2. 디렉터리형 패키지(Package)

- 패키지는 모듈을 모아 놓은 디렉토리 
- from 패키지명.하위패키지 import ...
- 전체 패키지 경로 : 절대 경로 접근 예) chapter06.mod1
- 상대 경로 접근법(..): 바로 위 패키지 
-               (.): 현재 모듈이 있는 패키지

### 3. import, from, as 키워드

- import : 모듈 불러오기 
- 불러온 모듈에서 마침표(.)를 통해서 접근
- from 모듈명/패키지명 import 모듈에 있는 함수,변수,클래스 등등,모듈
- from 모듈명 import * -> 모듈의 전체 함수, 변수, 클래스를 불러온다.
- as : alias - 별칭

```python
import mod1
```

```python
mod1.add(10, 20)
```

```python
mod1.VERSION
```

```python
import mod1 as m

m.add(10, 20)
```

```python
from mod1 import add, minus, VERSION
```

```python
add(10, 20)
```

```python
VERSION
```

```python
from mod1 import *
add(10, 20)
```

#### \_\_init\_\_.py 
- 현재 디렉토리가 패키지임을 알려주는 파일 
- 하위 모듈, 하위 패키지에서 공통적으로 처리해야 하는 코드를 작성, 모듈을 불러올때 최초 한번만 실행
- 패키지내에서 사용할 공통적인 함수, 변수, 클래스를 정의
- from 패키지 import * : \_\_all\_\_ = [....] : 에 정의된 모듈만 임포트


```python
from game import lol
```

```python
import game
```
