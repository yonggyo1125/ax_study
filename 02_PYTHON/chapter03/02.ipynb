## 3장 2강 : 매핑 및 집합 자료구조

### 1. 딕셔너리(Dictionary) 자료구조

- Key - Value 쌍 구조 
- Key : Value를 찾기 위한 키워드, 중복 X, 리스트는 사용 불가
- Value : Key로 찾는 값, 중복 O
- {"키": 값, "키": 값, ...}

- 추가(CREATE): 없는 키이름으로 값을 대입하면 추가
- 조회(READ): ['키 이름']
    - keys() : 모든 키를 담은 View, 읽기 전용
    - values() : 모든 값을 담은 View, 읽기 전용
    - items(): (키, 값) 목록을 담은 View

```python
user = {
        "email": "test@test.org",
        "password": 1234, 
        "name": "김철수"
    }
```

```python
user['mobile'] = "010-1000-1000"

user
```

```python
user['name']
```

```python
user.keys()
```

```python
user.values()
```

```python
user.items()
```
