### 2. 리스트(List)

- 변경 가능
- [값1, 값2, ...]

#### 주요 기능
추가(Create)
- append(요소): 끝에 추가 
- insert(추가 위치, 추가 요소)
- extend(콜렉션 - 리스트, 튜플): 원본 리스트 끝에 여러개 추가
- [기존 리스트] + [값1, 값2, ...] : 끝에 여러개 추가하고 새로운 리스트를 생성
,
조회(Read)
- 인덱스 번호로 접근 (fruits[1])
- index(요소): 특정 요소의 인덱스 번호

수정(Update)
- 인덱스 번호로 접근, 다른 값을 대입 (fruits[1] = '오렌지')

삭제(Delete)
- remove(요소) : 첫번째 발견한 특정 요소를 제거
- clear(): 리스트의 모든 요소 제거


기타 
- sort() : 정렬(오름차순), sort(reverse=True) : 내림차순
- reverse(): 순서 반전
- count(요소) : 특정 요소의 등장 횟수

```python
fruits = ['Apple', 'Orange', 'Banana']
```

```python
fruit1 = fruits[1]
fruit1
```

```python
fruits[1] = "오렌지"
fruits
```

```python
fruits.append("Melon")
fruits
```

```python
fruits_copy = fruits
fruits_copy[1] = 'Orange'
fruits_copy
```

```python
print(fruits)
print(fruits_copy)
```

```python
fruits.extend(["Cherry", "Water Melon"])
fruits
```

```python
fruits = ["Apple"]
print("추가 전(+):", id(fruits))

fruits = fruits + ["Orange", "Banana"]
print("추가 후(+):", id(fruits))

fruits
```

```python
fruits_extend = ["Apple"]
print("추가 전(extend):", id(fruits_extend))

fruits_extend.extend(["Orange", "Banana"])
print("추가 후(extend):", id(fruits_extend))
```

```python
fruits.sort(reverse=True) # 내림차순
fruits
```

### 3. 튜플(Tuple)

- 변경 불가 리스트 
- (값1, 값2, ....)
- (값1,) : 값이 하나인 튜플은 반드시 ,를 함께 추가
- 소괄호( )를 생략하고 콤마(,)로 값을 나열해서 정의 할수도 있음
- 읽기 전용
- 인덱스 연산 [] : 요소 접근 
- count(요소) : 요소의 등장 횟수
- index(요소) : 요소의 위치 번호

```python
fruits = ('Apple', 'Orange')
fruits
```

```python
fruits[0]
```

```python
fruits[0] = '사과'
```

```python
fruits2 = fruits + ('Banana', 'Melon')
fruits2
```

```python
print(id(fruits), id(fruits2))

print(fruits)
```

```python
fruits + ('Mango',)
```

```python
fruits = 'Apple', 'Orange', 'Banana'

fruits, type(fruits)
```

```python
apple = fruits[0]
orange = fruits[1]
banana = fruits[2]

apple, orange, banana
```

```python
apple, orange, banana = fruits

apple, orange, banana
```
