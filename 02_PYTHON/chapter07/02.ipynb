### 1.2 임시 print() 함수를 통한 디버깅과 한계

```python
total = 0
for i in range(100):
    total += i
    # 반복 횟수가 많아질수록 대량의 텍스트가 출력창을 오염시켜 흐름 식별을 방해함
    print(f"임시 로깅 수치 - i: {i}, 현재 total: {total}")
```

### 2.3 디버거 주요 흐름 제어 명령어의 특징

```python
def 가공_프로세스(n):
    return n * 2 # 'Step Into' 명령 수행 시에만 이 내부 라인으로 제어권이 진입합니다.

number = 10
# 'Step Over' 명령을 수행하면 함수 내부로 들어가지 않고 최종 연산 결과값만 도출하여 다음 줄로 넘어갑니다.
result = 가공_프로세스(number) 
print(result)
```

## 3. 트레이스백 분석 및 디버깅

### 3.1 에러 트레이스백 경로 추적 및 결함 지점 이동

```python
def calculate_average(scores):
    total = sum(scores)
    count = len(scores)
    average = total / count  # 결함 발생 예측 지점
    return average

def process_report():
    student_scores = []  # 비어있는 리스트 데이터 주입
    result = calculate_average(student_scores)
    print(f"평균 점수: {result}")

# 함수 호출을 통해 실행 흐름을 시작합니다.
process_report()
```

### 3.2 Debug Cell 활성화를 통한 단계별 변수 상태 검증

```python
# 실행 전, 4번째 라인인 'total += i' 왼쪽 공간을 클릭하여 빨간색 중단점(Breakpoint)을 설정합니다.
def accumulate_data():
    total = 0
    for i in range(1, 4):
        total += i  # [● 중단점 설정 위치 - 코드 왼쪽 여백 클릭]
    return total

# 셀 실행 버튼 옆의 드롭다운을 눌러 'Debug Cell'을 선택하여 디버거를 실행합니다.
final_result = accumulate_data()
print(f"최종 합계: {final_result}")
```
