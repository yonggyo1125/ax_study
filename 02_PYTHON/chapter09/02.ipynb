### 3. [데이터 표준 변환] json

```python
import json

# 파이썬 객체 -> JSON 문자열 
user = {
    "name": "김철수",
    "age": 20
}

json_string = json.dumps(user, ensure_ascii=False)
print("json_string:", json_string, "type:", type(json_string))
```

```python
# JSON 문자열 -> 파이썬 객체 
nums_string = "[1,2,3,4,5]"
user_string = '{"name": "김철수", "age": 20}'


nums = json.loads(nums_string)
user2 = json.loads(user_string)

print("nums:", nums, "type:", type(nums))
print("user2:", user2, "type", type(user2))
```
