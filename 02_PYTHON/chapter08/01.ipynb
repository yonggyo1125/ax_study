### 2. 클래스(Class)와 인스턴스(Instance)

```python
class Bus:

    # 속성 정의
    def __init__(self):
        self.count = 0 # 승객 수, 초기화: 변수가 되었을때 최초로 값을 대입
        self.passengers = []


    def ride(self, passenger): # 탑승
        self.count += 1 # 승객수 증가
        self.passengers.append(passenger)

        print("탑승")

    def off(self, passenger): # 하차
        if self.count > 1:
            self.count -= 1 # 승객수 차감
            self.passengers.remove(self.passenger)

        print("하차") 

    def show_count(self):
        print("승객 수:", self.count)

    def print_self(self):
        print(id(self))


class Passenger:
    pass
```

```python
bus1 = Bus()  # 객체, 클래스 -> 객체(인스턴스 - 실체)

# 마침표(.) 연산자를 통해서 객체의 (인스턴스)변수, 메서드에 접근 가능 

print("탑승객 수:", bus1.count)
bus1.ride()
```

```python
bus2 = Bus() 

bus1 is bus2 # 같은 클래스로부터 생성된 객체는 서로 다른 객체!
```

```python
bus1 = Bus()
print("bus1 주소:", id(bus1))

bus1.print_self()
```

```python
bus = Bus()
p1 = Passenger()
p2 = Passenger()

bus.ride(p1)
bus.ride(p2)

bus.show_count()
```

### 3. 초기화 함수 \_\_init\_\_ (생성자)

- 객체가 생성이 되면 가장 먼저 호출되는 메서드
- 객체가 생성된 직후에 해야할 처음 작업들(초기화 작업)
- 주로 하는 작업은 인스턴스 변수에 최초로 값을 할당(초기화)

### 4. self 키워드

- 클래스 내부에서 메서드의 첫번째 매개변수로 정의
- 객체가 생성이 되면 객체의 자원을 접근하기 위한 참조 변수
- 클래스 내부에서만 사용 가능

### 5. 속성(Attribute)과 인스턴스 메서드(Method)
