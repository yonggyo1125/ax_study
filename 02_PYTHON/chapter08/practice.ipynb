```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):

        minus_damage = damage if self.hp >= damage else self.hp
        self.hp -= minus_damage
        print("남은 체력:", self.hp)

        if self.hp <= 0: 
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")
```

```python
hero = Hero("피카츄")
hero.take_damage(90)
hero.take_damage(20)
```

```python
class DoorLock:
    def __init__(self, password):
        if not self.validate(password):
            return
        
        self.__password = str(password)

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if not self.validate(password):
            return
        
        self.__password = str(password)
     

    def validate(self, password):

        result = len(str(password)) >= 4
        if not result:
            print("경고: 비밀번호는 4자리 이상이어야 합니다.")

        return result
        
```

```python
door = DoorLock(123456)

door.__password = 'abcdef'
print(door.get_password())

door._DoorLock__password = 'abcdef'
print(door.get_password())
```

```python
class User:
    def __init__(self, name):
        self.name = name

class VIPUser(User):
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point
```

```python
user = VIPUser("김철수", 1000)

print(user.name, user.point)
```
