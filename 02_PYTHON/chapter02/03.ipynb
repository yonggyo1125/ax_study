## 2장 3강 : 연산자

### 1. 항과 연산자

- 항: 연산에 사용되는 값, 2개 -> 이항, 3개 -> 3항, 1개 -> 단항
- 연산자: 연산에 사용되는 기호(+, =, ...)

### 2. 연산자의 종류

#### 대입 연산자(=)
x = 10
x = 10 + 20

연산의 우선 순위가 가장 낮다. 가장 마지막에 연산된다.


#### 산술 연산자
+, -
* :곱하기
/ :나누기
// : 나눈 몫
% :나머지
** : 2**3 -> 2 *2 *2


- 연산 우선순위: **, *, /, //, % > +, -
- 우선 순위 강제 적용: (..)


```python
num1 = 10
num2 = 3

num1 // num2, num1 % num2
```

```python
num1 = 3

num1 ** 3
```

#### 비교 연산자
>, <, >=, <=
== : 같다
!= : 다르다 

- 연산의 결과는 무조건 True, False
- 판별이 필요한 구문 (조건문, 반복문)

```python
10 > 5
```
