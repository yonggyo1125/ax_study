```python
class DoorLock:
    def __init__(self, password):
        if not self.validate(password): return

        self.__password = str(password)

    def set_password(self, password):
        if not self.validate(password): return

        self.__password = str(password)
        
    def validate(self, password):
        result = len(str(password)) >= 4
        if not result: print("경고: 비밀번호는 4자리 이상이어야 합니다.")
        return result

    def __str__(self):
        return f"{self.__password}"
```

```python
d = DoorLock("00000")
d._DoorLock__password = "abcdf"

print(d)
```

```python
class User:
    def __init__(self, name):
        self.name = name


class VIPUser(User):
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point

    def __str__(self):
        return f"name={self.name}, point={self.point}"

    
```
