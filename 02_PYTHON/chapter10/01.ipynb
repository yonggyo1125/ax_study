### 1. with open()의 자원 자동 해제

with 구문: 첫 라인에서 \_\_enter\_\_(..)  실행, 마지막 라인 실행 후 \_\_exit\_\_(..)

```python
class MyResource:

    def __enter__(self):
        print("__enter__ 호출")

        return self

    # with 구문이 종료되었을 때 또는 예외가 발생했을때 호출
    # 예외가 발생시 exc_type - 예외 클래스 종류
    #             exc_val - 예외 메세지
    #             exc_tb - TraceBack 
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("__exit__호출")
```

```python
with MyResource() as o: # __enter__ 호출
    print(o)

    # 종료시 __exit__ 호출
```

```python
# 파일에서 작성하는 위치(파일 포인터)가 가장 앞으로 이동해서 내용을 대체
with open("test.txt", "w", encoding="UTF-8") as fp:
    fp.write("Hello!!")
```

```python
# 파일 포인터를 이동하지 않고 누적해서 작성 ( mode: a)
with open("test.txt", "a", encoding="UTF-8") as fp:
    fp.write("안녕하세요!\n")
```

### 2. 텍스트 파일 읽기 함수(read, readline, readlines)

- read(): 한꺼번에 전체 텍스트를 읽어온다.
- readline(): 한라인 읽어오기
- readlines(): 라인별로 읽어서 리스트로 반환

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    text = fp.read()
    print(text)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    # text = fp.readline()
    # print(text)
    # text = fp.readline()
    # print(text)
    # text = fp.readline()
    # print(text)
    # text = fp.readline()
    # print(text)
    while(text := fp.readline()): # 한줄씩 읽어서 메모리에 적재 하므로 메모리 소비가 적다
        print(text)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    lines = fp.readlines() # 한꺼번에 메모리에 올리기때문에 메모리 소비가 많다. 
    print(lines)
```

```python
with open("test.txt", "r", encoding="UTF-8") as fp:
    for line in fp:
        print(line)
```
