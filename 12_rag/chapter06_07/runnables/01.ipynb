```python
def func_a(text):
    return f"func_a:{text}"

def func_b(text):
    return f"func_b:{text}"
```

```python
from langchain_core.runnables import RunnableLambda

run_a = RunnableLambda(func_a)
run_b = RunnableLambda(func_b)

run_ab = run_a | run_b

run_ab.invoke("안녕하세요.")
```

```python
run_ab2 = run_a | func_b # __ror__( )

run_ab2.invoke("반갑습니다.")
```
