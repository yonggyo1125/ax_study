```python
from langchain_community.document_loaders import PyMuPDFLoader

pdf_path = "academic_rules_2026.pdf"
pdf_loader = PyMuPDFLoader(file_path=pdf_path)

pdf_documents = pdf_loader.load()

print(f"페이지 수: {len(pdf_documents)}")

for i, doc in enumerate(pdf_documents, 1):
    print(f"[{i}번째]: {doc}")
    print("-" * 50)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

rec_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

pdf_docs = rec_splitter.split_documents(pdf_documents)

for i, doc in enumerate(pdf_docs, 1):
    print(f"[{i}번째]: {doc}")
    print("-" * 50)
```

```python
from langchain_community.document_loaders import WebBaseLoader

target_url = "https://n.news.naver.com/mnews/article/001/0016439716"

web_loader = WebBaseLoader(web_path=target_url)

web_documents = web_loader.load()

print(f"페이지 수: {len(web_documents)}")

for i, doc in enumerate(web_documents, 1):
    print(f"[{i}번째]: {doc}")
    print("-" * 50)
```
