매개변수 기본값 (Default Parameter)

```javascript
function add(num1, num2 = 10) {
    return num1 + num2;
}

add(20);
```

arguments 객체

```javascript
// 함수 호출시 사용한 인수 목록
function sum() {
    let total = 0;
    for (const arg of arguments) {
        total += arg;
    }

    return total;
}
```

```javascript
console.log(sum(10, 20));
console.log(sum(10, 20, 30));
```

나머지 매개변수 (Rest Parameter, 가변 매개변수)

```javascript
// ES6+ 
// ... : 가변 매개변수
function sum(...params) {
    console.log(params);
}

sum(10, 20, 30, 40);
```

```javascript
function sum(num1, ...params) {
    console.log(num1, params);
}

sum(10, 20, 30, 40);
```

```javascript
// 전개 연산자 
function add(num1, num2) {
    return num1 + num2;
}

const nums = [10, 20];

console.log(add(nums[0], nums[1]));
console.log(add(...nums));
```

```javascript
const nums1 = [10, 20, 30, 40];
const nums2 = [...nums1];

console.log(nums2, nums1 === nums2);
```

```javascript
const person = {name: "김철수", age:20 };
const person2 = {...person};

console.log(person2, person === person2);
```

```javascript
const person3 = {...person, address: "주소"};
console.log(person3);
```

#### 비구조화 할당
- 구조 -> 비구조화(구조를 분해) -> 변수에 할당

```javascript
const nums = [10, 20, 30, 40, 50];

const [a, b, c] = nums;

console.log(a, b, c);
```
