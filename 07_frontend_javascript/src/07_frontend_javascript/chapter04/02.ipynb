```

```

#### 2.1 var의 3가지 결함
- 호이스팅 (Hoisting)
- 선언 키워드 생략 허용 (암묵적 전역 변수)

#### 6.1 조건문 (if, if-else, if-else if - else)

- 조건식 : 참, 거짓을 판별하는 연산 식(비교, 논리연산이 주로 쓰인다.)
```
if (조건식) {
    // 조건식 참일때 실행되는 구간

if (조건식) {
```

- 삼항 조건식
- 변수 = (조건식) ? 참일때 실행되는 코드 값 : 거짓일때 실행되는 코드나 값;

#### 6.2 선택문 (switch - case)

- switch ~ case : case 키워드에 매칭이되면 그게 바로 코드 실행 시작점이고 break를 만날때까지 실행
- break: 실행흐름을 중단 
- default: case에 매칭된 키워드가 없으면 마지막에 유입되는 구간

````javascript
const rank = 2;

switch(rank) {
    case 1:
        console.log("금메달");
        break;
    case 2:
        console.log("은메달");
        break;
    case 3:
        console.log("동메달");
        break;
    default:
        console.log("1,2,3등 중에서 입력")
}
````

````javascript
/*
const grade = 1~6;
1 - 1학년, 6 - 6학년, 1~6사이 값이 아니면 잘못 입력 출력
*/

const grade = 3;

switch(grade) {
    case 1: console.log("1학년"); break;
    case 2: console.log("2학년"); break;
    case 3: console.log("3학년"); break;
    case 4: console.log("4학년"); break;
    case 5: console.log("5학년"); break;
    case 6: console.log("6학년"); break;
    default: console.log("잘못 입력");
}
````

````javascript
// 2~9단 까지 중첩된 for문을 이용해서 작성하세요.
// 단보다 곱하는 수가 클때만 출력하세요!
for (let i = 2; i <= 9; i++) {
    console.log(`-------- ${i}단 ----------`);
    for (let j = 1; j <= 9; j++) {
        if (i >= j) continue; // 반복 건너띄기
        

        console.log(`${i} X ${j} = ${i*j}`);
    }
}
````

do-while 문 : 일단 한 번 실행하고 조건 비교하기

while 문

```
while(조건식) {
    // 조건식이 참일때 반복되는 구간
    // 반복을 중단 시키는 코드가 있어야 한다 - break와 함께 쓰인다.
}
```

do while 문
- 조건식이 거짓이더라도 한번 실행을 보장

```
do {
    
    // 조건식이 참일때 반복되는 구간

} while(조건식);
```

````javascript
let i = 101, total = 0;

while(i <= 100) {
    total += i;
    i++;
}

console.log("합계:", total);
````

````javascript
let i = 101, total = 0;

do {
    total += i;
    i++;

} while(i <= 100);

console.log("합계:", total);
````

### 7. 예외 처리 (try - catch - finally)

```
try {
    // 예외가 발생할 가능성이 있는 코드
```
