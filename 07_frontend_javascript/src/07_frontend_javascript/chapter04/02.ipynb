### 2. 초기 자바스크립트의 변수 선언: var의 치명적인 결함

```
var a = 100;

```
-> 번역 과정 -> window 객체의 하위 속성으로 추가(웹브라우저 기준) / global 객체의 하위 속성으로 추가(Node.js)

#### 2.1 var의 3가지 결함
- 호이스팅 (Hoisting)
- 선언 키워드 생략 허용 (암묵적 전역 변수)
- 함수 레벨 스코프 (Function-level Scope)

````javascript
var i = 10000; // window.i / global.i = 10000

for (var i = 0; i < 10; i++) { // window.i, global.i = 0 ~ 10
    // 반복 로직
    console.log("for문에 정의된 i:", i);
}

console.log(i); // window.i / global.i = 10
````

````javascript
console.log(a);

var a = 100;
````

````javascript
let i = 10000;

for (let i = 0; i < 10; i++) {
    // 반복 로직
}

console.log(i);
````

#### 3.2 let과 const의 구분 및 권장 사항
- let (재할당 필요 대상)
- const (재할당 불가 대상)
- 블록 스코프와 상수의 할당 검증

자바스크립트에서는 변수를 선언할 때 const로 선언하고 변경이 필요한 경우만 let으로 변경

````javascript
let c = 100;

c = 200;
console.log(c);
````

````javascript
const d = 100;
d = 200;
````

### 4. 데이터 타입과 typeof를 활용한 타입 검증

#### 4.1 자바스크립트의 주요 데이터 타입
- Number (숫자형) : 정수와 실수를 따로 구분하지 않고 모든 숫자 형식
- String (문자열) :'', ""로 감싼 텍스트
- Boolean (불리언) : true, false
- null: 값이 없는(비어 있는) 상태임을 표시하는 값
- undefined : 정의되지 않음, 변수를 선언 후 값을 대입하지 않으면 기본값

````javascript
const num1 = 100;

console.log(typeof num1); // number

const str1 = 'ABC'; // string
console.log(typeof str1);

const bool1 = true; // true, false
console.log(typeof bool1);
````

````javascript
let num2;

console.log(num2, typeof num2);
````
