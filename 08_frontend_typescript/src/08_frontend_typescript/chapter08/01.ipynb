## 8장 1강 : 속성 변경 및 가시성 제어 유틸리티

### 1. 속성 가시성 및 변경 제어 유틸리티

#### 1.1 선택적 속성으로 변환하는 Partial과 필수 속성으로 변환하는 Required

- `Partial<T>` : 타입 T의 모든 속성을 선택적 속성(?)으로 변경합니다.
- `Required<T>` : 타입 T의 모든 속성에서 선택적 수식어(?)를 제거하여 필수 속성으로 변경합니다.

```typescript
// 원본 사용자 프로필 인터페이스 정의
interface UserProfile {
  id: number; // 사용자 고유 식별자 (필수)
  name: string; // 사용자 이름 (필수)
  email: string; // 이메일 주소 (필수)
  age?: number; // 나이 (선택적 속성)
}
```

`Partial<T> 활용`

`Required<T> 활용`

#### 1.2 데이터를 안전하게 보호하는 Readonly
- `Readonly<T>`: 타입 T의 모든 속성을 읽기 전용(readonly)으로 변경합니다.
- 데이터 훼손 방지: 객체 생성 후 값이 임의로 수정되는 것을 컴파일 시점에 완벽히 차단합니다.

```typescript
// 원본 사용자 정보 인터페이스
interface UserConfig {
  theme: string; // 앱 테마 설정
  notifications: boolean; // 알림 수신 여부
}

```

### 2. 속성 추출 및 제외 유틸리티

#### 2.1 필요한 속성만 선택하는 Pick
- `Pick<T, K>`: 타입 T에서 지정한 속성 키 K만 골라내어 새로운 객체 타입을 생성합니다.
- 화면 맞춤형 타입 작성: 전체 회원 정보 중 마이페이지 요약 카드처럼 일부 정보만 표시할 때 사용합니다.

```typescript
// 원본 상품 정보 인터페이스
interface Product {
  id: number; // 상품 ID
  name: string; // 상품명
  price: number; // 가격
  description: string; // 상세 설명
  stock: number; // 재고 수량
}

// Product 타입에서 'name'과 'price' 속성만 선택하여 새로운 타입 생성


// 추출된 타입으로 객체 정의 (name과 price만 가짐)
```

#### 2.2 불필요하거나 민감한 속성을 제거하는 Omit
- `Omit<T, K>`: 타입 T에서 지정한 속성 키 K를 제외한나머지 속성들로 새로운 객체 타입을 생성합니다.
- 보안 및 자동 생성 필드 제외: 서버로 전송할 데이터에서 자동 생성되는 id나 민감 정보인 password를 제외할 때 사용합니다.

```typescript
// 회원가입용 전체 데이터 인터페이스
interface UserRegistration {
  id: number; // 시스템 자동 생성 ID
  username: string; // 아이디
  passwordHash: string; // 암호화된 비밀번호
  email: string; // 이메일
}

// UserRegistration 타입에서 'id'와 'passwordHash'를 제외한 타입 생성


// 제외된 필드를 뺀 입력 폼 전용 객체 생성
```
