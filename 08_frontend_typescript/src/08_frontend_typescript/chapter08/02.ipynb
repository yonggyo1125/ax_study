### 1. Record<K, T>를 활용한 키-값 타입 매핑
`Record<K, T>`는 정해진 키(K)들에 동일한 타입(T)의 값을 한 번에 짝지어 주는 타입 생성기입니다

#### 1.1 Record 타입
- 동적 속성 일괄 정의
> 개별적으로 키마다 타입을 선언하지 않고, 유니온 타입으로 선언된 키 집합 전체에 동일한 값 타입을 부여합니다.
- 속성 누락 방지
> 키 집합에 명시된 속성이 객체 생성 시 하나라도 빠지면 컴파일 단계에서 에러를 발생시켜 실수를 방지합니다.

과목별 성적 및 쇼핑몰 재고 관리

```typescript
// 과목 키 목록을 유니온 타입으로 정의합니다.
type Subject = "Korean" | "Math" | "English";

// Record<Key, Value>를 사용하여 각 과목이 점수(number)를 갖는 성적표 타입을 작성합니다.




console.log("학생 성적표 데이터:", myGrade);
// 쇼핑몰 카테고리별 재고 수량 관리 예시입니다.
type ProductCategory = "Electronics" | "Clothing" | "Food";

```

### 2. Exclude와 Extract를 이용한 유니온 타입 가공
Exclude는 불필요한 타입을 빼고, Extract는 필요한 타입만 골라내는 유니온 전용 필터입니다.

#### 2.1 Exclude<T, U>: 불필요한 타입 제외하기
유니온 타입 T에서 U에 해당하는 요소를 제거한 새로운 타입을 생성합니다.

#### 2.2 Extract<T, U>: 필요한 타입만 추출하기
유니온 타입 T에서 U와 공통되는 요소만 골라내어 새로운 타입을 생성합니다.

#### 2.3 Exclude & Extract 실습: 결제 수단 및 사용자 권한 가공

```typescript
// 전체 결제 수단 유니온 타입을 정의합니다.
type PaymentMethod = "CreditCard" | "BankTransfer" | "KakaoPay" | "CashOnDelivery";

// Exclude를 사용하여 현금 결제(CashOnDelivery)를 제외한 온라인 결제 수단 타입을 생성합니다.
```

```typescript
// 전체 사용자 권한 목록을 정의합니다.
type UserRole = "Admin" | "Manager" | "Student" | "Guest";

// Extract를 사용하여 관리자급 권한("Admin", "Manager")만 추출합니다.
```

### 3. ReturnType를 활용한 함수 반환 타입 역추적
ReturnType는 작성된 함수가 최종적으로 반환하는 값의 타입을 자동으로 계산해서 가져옵니다.

#### ReturnType의 필요성
- 코드 중복 제거: 함수의 리턴 데이터 구조가 변경되어도 ReturnType을 사용하면 관련 타입이 자동으로 업데이트됩니다.
- 외부 라이브러리 대응: 반환 타입이 별도로 export되어 있지 않은 외부 함수의 결과 타입을 역추출할 때 유용합니다.


사용자 프로필 생성 함수 데이터 추출

```typescript
// 사용자 프로필을 생성하여 반환하는 함수를 작성합니다.
function createStudentProfile(name: string, age: number, isEnrolled: boolean) {
  return {
    id: Date.now(),
    name: name,
    age: age,
    status: isEnrolled ? "Active" : "LeaveOfAbsence",
    createdAt: new Date().toISOString()
  };
}

// typeof 연산자와 ReturnType 유틸리티를 결합하여 함수의 반환 타입을 추출합니다.


```
