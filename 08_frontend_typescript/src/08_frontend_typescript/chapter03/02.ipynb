## 3장 2강 : 함수 표현식과 콜백 타입 정의

### 1. 인라인 함수 타입 표현식과 호출 시그니처
변수를 선언하는 시점에 함수의 매개변수와 반환 타입을 직접 지정하는 방식입니다.

#### 1.1 인라인 함수 타입 표현식 (Inline Function Type Expression)
화살표 함수 형태로 변수 선언부 자리에 `(매개변수: 타입) => 반환타입`을 작성하는 문법입니다.

```typescript
const vipPrice: (price: number) => number =  (x) => x * 0.9;
```

```typescript
type calVipPrice = (price: number) => number;

const vipPrice2: calVipPrice = (x) => x * 0.9;
```

#### 1.2 인라인 호출 시그니처 (Inline Call Signature)
- 중괄호(`{}`) 내부에 `(매개변수: 타입): 반환타입;` 형태로 함수의 구조를 정의하는 문법입니다.
- 화살표(`=>`) 대신 콜론(`:`)을 사용하며, 함수의 형태를 객체의 속성처럼 상세히 묘사할 때 활용합니다.

```typescript
const vipPrice3: {(price: number): number} = x => x * 0.9;
```

### 2. 고차 함수 인자로 전달되는 콜백 함수의 인라인 타입 정의
다른 함수를 매개변수로 받아서 내부에서 실행하는 함수를 고차 함수라고 합니다.

#### 2.1 배열 변환(map) 과정에서의 콜백 타입 지정

```typescript
type square = (score: number) => number;

const scores: number[] = [90, 80, 100, 60];

function convert(scores: number[], converter: square): number[] {
    const newScores = [];
    for (const score of scores) {
        const result = converter(score);
        newScores.push(result);
    }

    return newScores;
}
```

```typescript
convert(scores, x => x * x);
```
