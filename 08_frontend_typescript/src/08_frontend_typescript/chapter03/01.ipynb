## 3장 1강 : 매개변수와 반환 타입 명세

### 1. 함수의 입력과 출력 정의

#### 1.1 매개변수와 반환값 타입 명시
- 매개변수 타입: 함수가 전달받는 변수 이름 뒤에 `: 타입` 형태로 작성합니다.
- 반환 타입: 매개변수 소괄호 `()` 우측에 `: 타입` 형태로 작성하여 결과물의 형태를 제한합니다.

```typescript
function discountedPrice(amount: number, discountRate: number): number {
    const price = amount * (1 - discountRate);

    return price;
}
```

```typescript
discountedPrice(1000, 0.1);
```

```typescript
discountedPrice(1000);
```

### 2. 선택적 매개변수와 가변 인자 활용
상황에 따라 데이터가 입력되지 않을 수 있다면 선택적 매개변수나 가변 인자를 사용

#### 2.1 선택적 매개변수 (Optional Parameter)
- 필수 항목이 아닌 선택적 입력 항목을 처리할 때 변수명 뒤에 물음표(?)를 붙여 선언합니다.
- 선택적 매개변수 뒤에는 일반 필수 매개변수를 배치할 수 없으며, 인자가 전달되지 않으면 내부적으로 undefined 값을 가집니다.

```typescript
function discountedPrice(amount: number, discountRate?: number): number {
    console.log("amount:", amount, "rate:", discountRate);

    return 0;
}
```

```typescript
discountedPrice(1000, 0.1);
```

```typescript
discountedPrice(1000);
```

#### 2.2 가변 인자 (Rest Parameter)
매개변수의 개수가 고정되지 않고 여러 개가 들어올 때 사용하며, 항상 배열 타입(type[])으로 정의합니다.

```typescript
// 전달받은 점수들을 모두 더하는 함수입니다.
function aggregateScores(...scores: number[]): number {
    let total = 0;
    for (const score of scores) {
        total += score;
    }

    return total;
}

```

```typescript
aggregateScores(10, 20, 30, 40);
```
