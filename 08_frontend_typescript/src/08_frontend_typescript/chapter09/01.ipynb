### 3. 제네릭을 활용한 범용 응답 DTO 설계
모든 API 응답의 공통 껍데기(상태 코드, 성공 여부)를 제네릭 DTO로 재사용합니다.

#### 3.1 표준 API 응답 래퍼(Wrapper) DTO 구조화
- 공통 속성
> status(HTTP 상태 코드), success(성공 여부)
- 가변 속성
> data(실제 응답 데이터) 속성에 제네릭 T를 결합하여 어떤 데이터든 담을 수 있게 설계합니다.
