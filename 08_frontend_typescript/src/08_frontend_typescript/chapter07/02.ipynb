### 1. 추상 클래스(Abstract Class)
직접 객체(인스턴스)를 생성할 수는 없지만, 하위 클래스들이 반드시 구현해야 할 공통 표준 구조를 제공하는 클래스입니다.

#### 1.1 추상 클래스의 정의 및 특징
- abstract 키워드 : 클래스와 메서드 앞에 작성하여 추상 규격임을 선언합니다.
- 인스턴스화 제한: new 연산자를 사용해 직접 객체를 생성할 수 없습니다.
- 추상 메서드: 본문(`{}`)이 없으며, 상속받는 하위 클래스에서 반드시 재정의(오버라이딩)하여 구현해야 합니다.

```typescript
// 추상 클래스 선언: 직접 new 키워드로 생성할 수 없습니다.
abstract class ServiceUser {
    
}

// 추상 클래스를 상속받는 일반 사용자 클래스
class CustomerUser extends ServiceUser {
    
}

// 객체 생성 및 실행

```
