### 2. 클래스 상속과 super 키워드

#### 2.1 extends를 활용한 클래스 확장
- extends 키워드: 기존 부모 클래스의 기능을 그대로 유지하면서, 새로운 기능을 추가한 자식 클래스를 생성합니다.
- 코드 재사용: 중복된 코드 작성을 방지하고, 유지보수 효율성을 높입니다.

#### 2.2 부모 생성자 호출 및 자원 접근(super)과 메서드 오버라이딩
- `super()` 생성자 호출: 자식 클래스의 생성자에서 부모 클래스의 생성자를 실행할 때 사용하는 명령어입니다.
- `super.메서드 접근`: 자식 클래스 내부에서 부모 클래스에 정의된 기존 메서드나 자원에 직접 참조로 접근할 때 사용합니다.
- 메서드 오버라이딩(Overriding): 부모 클래스로부터 물려받은 메서드를 자식 클래스의 목적에 맞게 재정의하는 작업입니다.
