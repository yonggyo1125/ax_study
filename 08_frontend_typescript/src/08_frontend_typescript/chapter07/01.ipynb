## 7장 1강 : 클래스 구조와 접근 제한자 및 상속

### 1. 클래스 구조와 접근 제한자
- 클래스(Class): 객체를 찍어내는 틀
- 접근 제한자: 내부 데이터를 외부에 얼마나 공개할지 결정하는 안전장치입니다.

#### 1.1 TypeScript 클래스 필드 선언과 생성자
- 클래스 필드(Field) : 클래스 내부에서 관리되는 변수이며, TypeScript에서는 타입을 미리 명시해야 합니다.
- 생성자(Constructor) : 객체가 생성될 때 가장 먼저 실행되는 함수로, 필드 값을 초기화하는 역할을 수행합니다.

```typescript
// 은행 계좌를 만드는 클래스 정의
class BankAccount {
    // public owner: string;
    // public balance: number;
    // owner: string; // public
    // balance: number; // public
    private owner: string;
    private balance: number;

    constructor(owner: string, balance: number) {
        this.owner = owner;
        this.balance = balance;
    }

    getSummary(): string {
        return `소유자: ${this.owner} / 잔액: ${this.balance}`;
    }
}

const myAccount = new BankAccount("김철수", 10000);
console.log(myAccount.getSummary());
console.log(myAccount.owner);
```

#### 1.2 접근 제한자로 데이터 보호하기 (public, protected, private)
- public: 클래스 내부, 자식 클래스, 외부 어디서든 접근이 가능합니다. (기본 설정값)
- protected: 클래스 내부와 상속받은 자식 클래스에서만 접근이 가능하며, 외부에선 접근이 불가능합니다.
- private: 해당 클래스 내부에서만 접근이 가능하며, 자식 클래스나 외부에서는 완전히 차단됩니다

```typescript
// 접근 제한자가 적용된 계좌 클래스
class SecureAccount {
    public owner: string;
    protected accountType: string; // 클래스 내부만 가능 + 상속을 통해서 하위 클래스 내부에서 가능
    private balance: number;

    constructor(owner: string, accountType: string, balance: number) {
        this.owner = owner;
        this.accountType = accountType;
        this.balance = balance;
    }
}

const myAcc = new SecureAccount("김철수", "보통예금", 10000);
console.log("owner:", myAcc.owner);
// console.log("accountType:", myAcc.accountType);
```

```typescript
class SubSecureAccount extends SecureAccount {
    constructor(owner: string, accountType: string, balance: number) {
        super(owner, accountType, balance);
    }

    // getBalance(): number {
    //     return super.balance;
    // }

    getType(): string {
        return super.accountType;
    }
}

const myAcc2 = new SubSecureAccount("이영희", "정기적금", 50000);
// console.log("accountType:", myAcc2.accountType);
console.log("accountType:", myAcc2.getType());
```

#### 1.3 생성자 단축 속성 선언
생성자 매개변수 앞에 접근 제한자를 작성하면, 필드 선언과 this 할당 과정을 자동으로 처리합니다.

```typescript
// 단축 속성 선언을 활용하여 간결하게 작성한 클래스
class CompactAccount {
    constructor(public owner: string, protected accountType: string, private balance: number) {}

    public getSummary(): string {
        return `소유자:${this.owner}, 계좌 타입: ${this.accountType}, 잔고: ${this.balance}`;
    }
}

const myAcc3 = new CompactAccount("김철수", "보통예금", 20000);
console.log(myAcc3.getSummary());
```
