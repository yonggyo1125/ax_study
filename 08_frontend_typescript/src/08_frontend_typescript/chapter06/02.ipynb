### 2. keyof 연산자의 이해와 활용
객체 타입이 가진 모든 속성 키(Key)의 이름을 문자열 리터럴 유니온 타입으로 추출하는 연산자입니다.

#### 2.1 keyof 연산자의 기본 문법
- 객체 형태의 타입 앞에 작성하여, 해당 객체의 속성명들을 유니온 타입으로 추출합니다.
- 오타 등으로 인해 존재하지 않는 속성에 접근하는 실수를 방지합니다.

```typescript
// 사용자 프로필 정보를 나타내는 인터페이스를 선언합니다.
interface UserProfile {
    id: number;
    name: string;
    email: string;
    age: number;
}

// keyof 연산자로 UserProfile의 속성 키만 추출합니다.
```
