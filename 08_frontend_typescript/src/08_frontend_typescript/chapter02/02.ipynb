## 2장 2강 : 특수 타입과 타입 추론 메커니즘

### 1. 특수 타입 (any, unknown, never)
> 특정한 상황이나 형식이 명확하지 않은 데이터를 처리하기 위해 정의된 타입 규격입니다.

#### 1.1 타입 검사를 제외하는 any 타입
- 타입스크립트의 컴파일 시점 정적 타입 검사를 수행하지 않도록 설정하는 타입입니다.
- 모든 종류의 데이터를 대입할 수 있으나, 컴파일러가 타입 오류를 감지하지 않으므로 실행 시점에 오류가 발생할 수 있습니다.

```typescript
let flexibleValue: any = "정산 완료";
```

#### 1.2 타입 검증과 타입 좁히기를 강제하는 unknown 타입
- any와 동일하게 모든 종류의 데이터를 대입할 수 있지만, 정체를 확인하기 전에는 사용을 제한하는 안전한 타입입니다
- 외부 API의 응답 데이터처럼 형식을 미리 알 수 없는 데이터를 다룰 때 컴파일 시점의 안정성을 보장하기 위해 사용합니다
- 조건문을 통해 컴파일러에게 특정 변수의 타입을 명확히 입증하여, 해당 블록 안에서 안전하게 데이터에 접근할 수 있도록 허용하는 과정입니다.


| **검증 방식** | **대상 데이터** | **설명** |
| --- | --- | --- |
| typeof 연산자 | 원시 타입 | string, number, boolean 등 기본 데이터 타입을 판별할 때 사용합니다. |
| Array.isArray() | 배열 구조 | 데이터가 배열 형식인지 검증할 때 사용합니다. |
| 사용자 정의 타입 가드 | 객체 및 커스텀 타입 | is 키워드를 사용하여 객체의 내부 속성 구조를 직접 검증할 때 사용합니다. |

```typescript
let rawResponse: unknown;
```

typeof 연산자를 활용한 원시 타입 좁히기

```typescript
rawResponse = "   TypeScript  ";
```

Array.isArray()를 활용한 배열 타입 좁히기

```typescript
rawResponse = [10, 20, 30];
```

#### 1.3 any보다 unknown 사용을 지향해야 하는 이유

| **비교 항목** | **any 타입** | **unknown 타입** |
| --- | --- | --- |
| **값의 할당** | 모든 타입의 값을 대입할 수 있음 | 모든 타입의 값을 대입할 수 있음 |
| **속성 및 메서드 접근** | 타입 검사 없이 즉시 접근 가능 | 타입 좁히기 전에는 접근 불가능 (컴파일 에러) |
| **안전성 수준** | 낮음 (자바스크립트와 동일하게 오류 위험 노출) | 높음 (컴파일러가 사전 검증을 강제함) |

any 사용 시의 문제점
> any를 사용하면 타입스크립트의 정적 분석 기능이 무력화되므로, 오타나 잘못된 메서드 호출을 컴파일 시점에 잡아낼 수 없습니다. 따라서 정체를 알 수 없는 데이터는 우선 unknown으로 선언한 뒤 타입 좁히기를 통해 안전하게 구체적인 타입을 찾아내는 것이 올바른 설계 방향입니다.

#### 1.4 값을 반환하지 않는 never 타입
- 값을 가질 수 없으며, 함수의 실행이 정상적으로 종료되지 않음을 의미하는 타입입니다.
- 의도적으로 예외 오류를 발생시키거나 무한 루프를 실행하여 반환값 자체가 존재하지 않는 구조에 지정됩니다.

예외 발생 및 종료되지 않는 구조 명세

```typescript
function triggerSevereError(message: string): never {
    // 함수가 값을 반환하며 종료되지 않고 예외를 발생시킵니다.
    throw new Error(message); 
}

function processInfiniteLoop(): never {
    // 조건이 참으로 유지되어 종료 지점에 도달하지 않는 함수입니다.
    while (true) {
        console.log("프로세스 실행 중");
    }
}
```

#### 1.5 타입스크립트 타입 계층 구조
- 최상위 타입 (Top Type) : unknown
<br>모든 데이터 타입을 포함할 수 있는 상위 계층의 타입

- 최하위 타입 (Bottom Type):  never
<br>어떤 값도 포함하지 않는 하위 계층의 타입

### 2. 타입 추론 (Type Inference)
코드를 작성할 때 타입을 직접 명시하지 않아도 대입된 초기값을 기준으로 타입을 자동 지정하는 메커니즘입니다.

#### 2.1 초기값에 의한 타입 확정 규칙

```typescript
let regularPrice = 45000; // 숫자가 대입되었으므로 number 타입으로 추론됩니다.
let accountId = "user_7789"; // 문자열이 대입되었으므로 string 타입으로 추론됩니다.

// 정해진 타입과 다른 형식의 데이터를 대입하려고 하면 컴파일 에러가 발생합니다.
```

### 3. 타입 단언 (Type Assertion)
- 컴파일러가 유추하는 타입보다 개발자가 해당 데이터의 타입을 더 명확히 알고 있을 때, 컴파일러에게 특정 타입으로 간주하도록 직접 지시하는 구문입니다.
- 형 변환(Type Casting)과 달리 실제 실행 시점(Runtime)의 데이터를 변환하지 않으며, 오직 컴파일 시점의 타입 검사만 재정의합니다.

#### 3.1 as 키워드와 타입 단언
> 변수나 표현식 뒤에 as 타입명을 붙여 컴파일러에게 데이터의 타입을 지정할 수 있습니다.

```typescript
let unknownData: unknown = "TypeScript 5.0";

// 컴파일러는 unknownData의 속성을 모르지만, 개발자가 string임을 확신하여 단언
let strLength: number = (unknownData as string).length;

console.log("문자열 길이:", strLength);
```

#### 3.2 타입 단언 활용 (JSON 및 API 데이터)

```typescript
interface UserProfile {
    id: number;
    name: string;
    role: string;
}

// 외부에서 전달받은 JSON 문자열 데이터
const rawJsonData = '{"id": 101, "name": "김스파르타", "role": "admin"}';

// JSON.parse() 결과를 UserProfile 타입으로 단언
const user = JSON.parse(rawJsonData) as UserProfile;


// 타입 단언을 통해 자동 완성 및 속성 접근 지원
console.log("사용자 ID:", user.id);
console.log("사용자 이름:", user.name);
console.log("권한:", user.role);
```

#### 3.3 타입 단언 사용 시 주의사항
> 타입 단언은 컴파일러의 타입 검사기를 강제로 우회하는 기능이며 자바스크립트 변환 시 삭제됩니다. 실제 데이터가 숫자인데 as string으로 단언하더라도 실행 시점에 숫자가 문자열로 바뀌지 않습니다. 따라서 외부 데이터의 형식을 100% 확신할 수 있는 상황에서만 제한적으로 사용해야 합니다.

```typescript
// 잘못된 타입 단언 예시 (컴파일은 통과하지만 런타임 오류 발생)
let rawNumber: unknown = 12345;

// 실제 값은 숫자이지만 개발자가 string으로 잘못 단언 (컴파일 타임에 감지 못 함)
let wrongAssertedStr = rawNumber as string;

// 실행 시점에 toUpperCase() 함수가 없어 런타임 에러 발생!
console.log(wrongAssertedStr.toUpperCase()); // TypeError: wrongAssertedStr.toUpperCase is not a function
```
