## 4장 1강 : 타입 별칭과 인터페이스 및 readonly

### 1. 타입 별칭과 인터페이스
- 타입 별칭과 인터페이스는 객체의 '설계도'를 그리는 문법입니다.
- 자바스크립트의 객체가 어떤 모양의 데이터(문자열, 숫자 등)를 가져야 하는지 미리 규격을 정해두는 역할을 합니다.

#### 1.1 타입 별칭(Type Alias)으로 객체 정의하기
- type 키워드를 사용하여 기존 타입에 새로운 이름을 붙여주는 방식입니다.
- 객체뿐만 아니라 단순한 문자열이나 숫자 배열 등 모든 타입을 가리킬 수 있습니다.

```typescript
// 유저 정보의 규격을 정의하는 타입 별칭 선언
type UserType = {
    id: number;
    name: string;
    email: string
}

// 정의된 설계도를 바탕으로 실제 객체 데이터 생성


// 데이터 출력하여 확인하기
```

#### 1.2 인터페이스(Interface)로 객체 정의하기
- interface 키워드를 사용하여 오직 객체의 구조만을 정의하기 위해 만들어진 문법입니다.
- 상속이나 확장성이 뛰어나 리액트(React) 등 웹 프레임워크에서 데이터를 주고받을 때 자주 쓰입니다.

```typescript
interface ProductInterface {
    code: string;
    price: number;
    brand: string;
}


// 인터페이스 규격을 따르는 실제 객체 데이터 생성


// 데이터 출력하여 확인하기
```

#### 1.3 타입 별칭과 인터페이스의 차이점

| **구분 (기능 및 특징)** | **인터페이스 (interface)** | **타입 별칭 (type alias)** |
| --- | --- | --- |
| **객체(object) 구조 정의** | ✅ 가능 | ✅ 가능 |
| **객체 구조 확장 (extends)** | ✅ extends 사용 | ✅ 인터섹션(`&`) 사용 |
| **클래스 구현 (implements)** | ✅ 가능 | ✅ 가능 |
| **선언 병합 (Declaration Merging)** | ✅ 가능 | ❌ 불가능 |
| **복합 타입 정의 (유니온 / 인터섹션)** | ❌ 불가능 | ✅ `\|`, `&`로 가능 |
| **함수(function) 타입 정의** | ❌ 불가능 | ✅ 가능 |
| **객체 이외의 다양한 타입 표현 (원시값 등)** | ❌ 불가능 | ✅ 가능 |
| **typeof 연산자를 활용한 타입 추출** | ❌ 불가능 | ✅ 가능 |

### 2. 속성 제어자: readonly와 선택적 속성(?)
- readonly와 물음표(?)는 데이터의 수정 가능 여부와 필수 여부를 제어합니다.
- 객체 속성의 안전성을 보장하거나 불필요한 에러 없이 유연한 데이터 처리를 가능하게 합니다.

#### 2.1 readonly로 데이터 훼손 방지하기
- 객체 생성 후 특정 속성의 값을 수정할 수 없도록 '읽기 전용'으로 잠그는 키워드입니다.
- 계좌번호, 결제 금액, 고유 ID 등 시스템 내에서 변경되지 않아야 할 필수 데이터를 보호합니다.

```typescript
// 계좌 정보를 정의하는 인터페이스
interface Account {
    readonly accountId: string; // 읽기 전용 속성 (생성 후 변경 불가)
    owner: string;
    balance: number;
}

// 계좌 객체 생성
const myAccount: Account = {
    accountId: "110-123-456789",
    owner: "김철수",
    balance: 50000
};

// 일반 속성은 정상적으로 수정이 가능합니다.


// [컴파일 에러] readonly 속성을 변경하려고 시도하면 컴파일 단계에서 차단됩니다.
```

#### 2.2 선택적 속성(?)으로 유연한 객체 설계하기
- 속성명 뒤에 물음표(?)를 붙여 데이터 입력의 선택 가능성을 지정하는 제어자입니다.
- 프로필 자기소개, 선택 옵션 등 값이 없을 수도 있는 상황에서 에러를 방지합니다.
