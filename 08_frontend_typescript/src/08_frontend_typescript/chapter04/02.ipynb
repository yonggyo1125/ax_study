## 4장 2강 : 인덱스 시그니처와 인터페이스 확장

### 1. 동적 속성을 다루는 인덱스 시그니처
인덱스 시그니처(Index Signature)
> 인덱스 시그니처는 속성의 이름(Key)을 지정하지 않고, Key와 Value의 타입 규격만 정의하여 동적 객체를 다루는 문법입니다.

#### 1.1 사전에 키를 알 수 없는 동적 객체 타입 정의

인덱스 시그니처
> 객체에 들어올 속성 이름을 미리 확정할 수 없을 때 [key: T]: V 형태로 규격을 지정합니다.

허용되는 Key 타입
> Key 위치에는 `string, number, symbol`, 그리고 특정 패턴을 가진 문자열인 `템플릿 리터럴 타입`을 지정할 수 있습니다.

Key 타입의 제약
> 인덱스 시그니처 Key 타입에는 유니온 타입('id' | 'name')을 직접 지정할 수 없습니다. 특정 키 값으로만 제한하려면 `Record<K, T>` 타입이나 `맵드 타입(Mapped Type)`을 사용해야 합니다.

string 및 number 키 타입 정의

```typescript
interface SubjectScores {
    [subjectName: string]: number; // 속성 이름은 string 타입, 값은 number 타입으로 지정
}

```

symbol 키 타입 정의 (TypeScript 4.4+)

```typescript
const idSymbol: unique symbol = Symbol("id");
interface SymbolKeyObject {
    [key: symbol]: string; // 키로 symbol 타입을 허용
}
```

템플릿 리터럴 키 타입 정의 (TypeScript 4.4+)

```typescript
interface ApiEndpoints {
    [key: `get_${string}`]: string; // 키가 반드시 'get_'으로 시작하는 문자열이어야 함
}

```

특정 유니온 키 제한 시 대안 (Record 활용)

```typescript
type AllowedKeys = "id" | "name";
```

### 2. 인터페이스 확장과 선언 병합
인터페이스 확장은 기존 정의된 속성을 바탕으로 새로운 속성을 추가하여 타입을 재사용하는 방법입니다.

#### 2.1 extends 키워드를 활용한 상속 구조

```typescript
// 기본 상품 정보를 정의하는 부모 인터페이스
interface Product {
  id: number;     // 상품 식별 번호
  name: string;   // 상품 이름
  price: number;  // 상품 가격
}

// Product를 상속받아 전자제품 규격을 정의하는 자식 인터페이스
interface Electronics extends Product {
  warrantyMonths: number; // 보증 기간(월) 속성 추가
}

// 확장된 Electronics 인터페이스를 적용한 객체 생성
```

#### 2.2 자동 선언 병합 (Declaration Merging)
- 동일한 이름으로 선언된 여러 인터페이스는 TypeScript 컴파일러에 의해 하나로 합쳐집니다.
- 동일한 인터페이스 이름으로 작성된 속성들이 하나의 타입 구조로 병합됩니다.

```typescript
// 첫 번째 User 인터페이스 선언
interface User {
    id: number;  // 사용자 식별 번호
    name: string; // 사용자 이름
}

// 동일한 이름으로 두 번째 User 인터페이스 선언
interface User {
    email: string; // 이메일 속성
}

// 병합되어 id, name, email 속성이 모두 지정된 객체 생성
```
