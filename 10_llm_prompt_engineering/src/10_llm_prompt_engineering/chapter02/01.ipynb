## 2장 1강 : System·User·Assistant 메시지 구조

### 3. 역할별 메시지 템플릿 조립하기

- System: 모델의 역할(페르소나), 상황, 과업, 규칙 정의, 최우선순위 / SystemMessage
- User: 실제 질의 응답에 대한 사용자 질문, 처리할 데이터의 전달, 가변적, 프롬프트 템플릿 / HumanMessage
- Assistant:  AI가 출력한 응답 / AIMessage

- User - Assitant -> 대화의 맥락 정보를 쌍으로 전달, 1쌍 - 턴, 여러쌍 -> 멀티 턴
- BaseMessage의 하위 클래스(HumanMessage, SystemMessage, AIMessage, ToolMessage)

#### 3.1 라이브러리 로드 및 환경 설정

````python
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage, BaseMessage

model = init_chat_model(
    model_provider="ollama",
    model="mistral"
)

parser = StrOutputParser()

chain = model | parser
````

#### 3.2 역할별 메시지 템플릿 생성 및 실행

````python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "당신은 재치있게 컴퓨터의 개념을 설명하는 유치원 교사입니다."),
    ("user", "다음 개념을 설명하세요: {text}")
])

chain2 = prompt_template | model | parser

response = chain2.invoke({"text": "함수"})

response
"""
' 아싸! 컴퓨터는 매우 중요한 도구입니다. 오늘날, 우리는 컴퓨터를 일상 생활에서 굉장히 많이 사용합니다. 예를 들어, 전화, 텔레비전, 게임 등. 오늘은 함수(Function)에 대해 알아보겠습니다.\n\n함수는 컴퓨터 프로그래밍에서 중요한 개념입니다. 함수는 특정 작업을 수행하는 명령의 모음을 의미합니다. 예를 들어, 두 수를 더하거나 뺄 때, 항상 같은 일련의 단계(코드)를 반복해야 하는 것이 아닌 한, 함수를 만들어 놓고 필요할 때 그 함수를 사용하면 됩니다! 이렇게 함수를 사용하면 코드를 간결하고 효율적으로 만들 수 있습니다!\n\n함수를 정의하는 방법은 다음과 같습니다:\n\n1. 함수 이름 선언: 함수의 이름을 선언합니다. 예: `add()`\n2. 파라미터 선언: 함수에 전달될 인수를 정의합니다. 예: `(num1, num2)`\n3. 인덱스 매개변수 선언: 인수를 저장할 변수를 선언합니다. 예: `a = num1, b = num2`\n4. 함수 몸체: 함수 내부에서 실행할 코드를 작성합니다. 예: `result = a + b`\n5. 반환 키워드: 함수가 수행된 결과를 반환합니다. 예: `return result`\n6. 함수 종료: 함수가 수행되는 코드 블록을 닫습니다.\n\n이렇게 정의된 함수를 사용하려면 다음과 같이 합니다:\n\n```python\ndef add(num1, num2):\n    a = num1\n    b = num2\n    result = a + b\n    return result\n\nresult = add(3, 4) # 결과: 7\n```\n\n함수는 매우 중요한 개념입니다. 함수를 사용하면 코드를 효율적으로 작성하고 재사용하는 것이 가능해집니다. 앞으로도 함수를 자주 사용하고 이해하시면 됩니다!\n\n[함수 학습 사이트](https://www.w3schools.com/python/python_functions.asp)에서 더욱 자세한 학습 자료를 확인하시면 됩니다!'
"""
````

### 4. 대화 이력(Memory) 구조 설계하기

#### 4.1 Assistant 메시지 기반 문맥 유지

````python
messages: list[BaseMessage] = [
    HumanMessage("내 이름 김철수야. 만나서 반가워"),
    AIMessage("반갑습니다. 철수님, 무엇을 도와드릴까요?"),
    HumanMessage("오늘 점심 메뉴를 추천해줘"),
    AIMessage("오늘 점심은 돈까스가 좋을것 같아요."),
    HumanMessage("오늘 점심 추천이 뭐었지?")
    #HumanMessage("내 이름이 뭐라고 했지?")
]

response = chain.invoke(messages)
response
````
