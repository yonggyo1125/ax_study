### 2. 명확한 지시의 4대 핵심 구성 요소

- **임무 (Task)**: AI가 수행해야 하는 핵심 동작을 명확한 동사로 표현합니다.
- **맥락 (Context)**: 작업의 배경, 목적, 사용 환경(예: 웹 서비스 화면)을 설명합니다.
- **제약 조건 (Constraints)**: 대상 독자, 글자 수, 문체, 금지 사항 등 지켜야 할 규칙을 지정합니다.
- **출력 형식 (Format)**: 마크다운, 리스트, 표 등 결과물이 출력되어야 하는 형식을 지정합니다.
