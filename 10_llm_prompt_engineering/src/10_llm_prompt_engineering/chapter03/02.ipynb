## 3장 2강 : One-shot과 Few-shot 프롬프팅

### 3. One-shot 및 Few-shot 프롬프트 설계 실습

#### 3.1 문체 변환을 위한 One-shot 프롬프트 실습

system으로 역할을 부여하고 첫 번째 user-assistant 쌍으로 단 1개의 모범 예시를 전달하여 일상 대화 문장을 '공식적인 비즈니스 이메일 문체'로 변환합니다.

#### 3.2 고객 문의 분류를 위한 Few-shot 프롬프트 실습

각 카테고리별로 1개씩 동일한 비율의 예시 데이터를 배치하여 모델의 중립성을 확보하고, 고객 문의 데이터를 '배송', '품질', '결제'의 3가지 유형으로 분류하는 다중 분류 체인을 구현합니다.
